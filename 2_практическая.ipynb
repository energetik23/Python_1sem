1. На входе две переменные employed и vacation (оба булевого типа). Необходимо определить нужно ли пользователю поставить будильник на утро. Только в случае если пользователь трудоустроен и не в отпуске нужен будильник. Вернуть булево значение.

```python
# Задание 1
def need_alarm(employed: bool, vacation: bool) -> bool:
    """
    Будильник нужен только если пользователь трудоустроен И не в отпуске.
    """
    return employed and not vacation


# Проверка
print(need_alarm(True, False))   # True  — работаем, не в отпуске
print(need_alarm(True, True))    # False — в отпуске
print(need_alarm(False, False))  # False — не трудоустроен
```

2. На входе три переменные operation (строка), value_1 и value_2 (целые числа). Необходимо выполнить вычисление в зависимости от выбранной математической операциии.
('+', 4, 7) --> 11
('-', 15, 18) --> -3
('*', 5, 5) --> 25
('/', 49, 7) --> 7

```python
# Задание 2
operation = input("Операция (+, -, *, /): ")
value_1 = int(input("Первое число: "))
value_2 = int(input("Второе число: "))

if operation == '+':
    print(value_1 + value_2)
elif operation == '-':
    print(value_1 - value_2)
elif operation == '*':
    print(value_1 * value_2)
elif operation == '/':
    if value_2 != 0:
        print(value_1 / value_2)
    else:
        print("Деление на ноль!")
else:
    print("Неизвестная операция")
```

3. Даны два числа. Посчитать сумму всех чисел между ними включительно.

```python
# Задание 3
a = int(input("Первое число: "))
b = int(input("Второе число: "))
start = min(a, b)
end = max(a, b)
total = 0
for i in range(start, end + 1):
    total += i
print("Сумма:", total)
```

4.  На вход подается номер месяца. Определить к какому кварталу и какому сезону относится месяц.

```python
# Задание 4
month = int(input("Номер месяца (1-12): "))

if month in (1, 2, 3):
    quarter = 1
elif month in (4, 5, 6):
    quarter = 2
elif month in (7, 8, 9):
    quarter = 3
elif month in (10, 11, 12):
    quarter = 4
else:
    quarter = 0
    print("Некорректный номер месяца")

if month in (12, 1, 2):
    season = "Зима"
elif month in (3, 4, 5):
    season = "Весна"
elif month in (6, 7, 8):
    season = "Лето"
elif month in (9, 10, 11):
    season = "Осень"
else:
    season = "-"

if quarter != 0:
    print(f"{quarter}-й квартал, сезон — {season}")
```

5. Посчитать средний балл из оценок, вводимых с клавиатуры, оценки вводятся бесконечно до момента ввода числа -1. Вернуть оценку буквой
	90 и выше - A,
	80-89 - B,
	70-79 - C,
	60-69 - D,
	Ниже 60 - F

```python
# Задание 5
total = 0
count = 0

while True:
    grade = int(input("Введите оценку (-1 для завершения): "))
    if grade == -1:
        break
    total += grade
    count += 1

if count == 0:
    print("Оценок нет")
else:
    average = total / count
    if average >= 90:
        letter = 'A'
    elif average >= 80:
        letter = 'B'
    elif average >= 70:
        letter = 'C'
    elif average >= 60:
        letter = 'D'
    else:
        letter = 'F'
    print(f"Средний балл: {average:.2f}, оценка: {letter}")
```
