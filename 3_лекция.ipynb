**Find maximum and minimum of values in list**

Ваша задача - создать две функции ( max и min, или maximum и minimum и т.д., В зависимости от языка), которые получают список целых чисел в качестве входных данных и возвращают наибольшее и наименьшее число в этом списке соответственно. Каждая функция возвращает одно число.

Примеры (ввод -> вывод)
* [4,6,2,1,9,63,-134,566]         -> max = 566, min = -134
* [-52, 56, 30, 29, -54, 0, -110] -> min = -110, max = 56
* [42, 54, 65, 87, 0]             -> min = 0, max = 87
* [5]                             -> min = 5, max = 5

Примечания

Можно считать, что пустых массивов/векторов не будет.

```python
def minimum(arr):
    return min(arr)


def maximum(arr):
    return max(arr)
```

**Word values**

Если дана строка "abc" и предполагается, что каждая буква в строке имеет значение, равное ее порядковому номеру в алфавите, то наша строка будет иметь значение 1 + 2 + 3 = 6. Это означает, что: a = 1, b = 2, c = 3 ... z = 26.

Вам будет дан список строк, и ваша задача — вернуть значения строк, как описано выше, умноженные на порядковый номер этой строки в списке. Для наших целей порядковый номер начинается с 1.

["abc", "abc abc"] должно возвращать [6, 24] из-за [ 6 * 1, 12 * 2 ]. Обратите внимание, что пробелы игнорируются.

"abc" имеет значение 6, а "abc abc" — значение 12. Теперь значение в позиции 1 умножается на 1, а значение в позиции 2 — на 2.

Входные данные будут содержать только строчные буквы и пробелы.

```python

def name_value(my_list):
    result = []
    for i, word in enumerate(my_list, start=1):
        value = sum(ord(char) - ord('a') + 1 for char in word if char != ' ')
        result.append(value * i)
    return result
```

**Remove all marked elements of list**

Определите метод / функцию, которая удаляет из данного массива целых чисел все значения, содержащиеся во втором массиве.

Примеры (ввод -> вывод):
* [1, 1, 2, 3, 1, 2, 3, 4], [1, 3] -> [2, 2, 4]
* [1, 1, 2, 3, 1, 2, 3, 4, 4, 3, 5, 6, 7, 2, 8], [1, 3, 4, 2] -> [5, 6, 7, 8]
* [8, 2, 7, 2, 3, 4, 6, 5, 4, 4, 1, 2, 3], [2, 4, 3] -> [8, 7, 6, 5, 1]

```python
class List:
    def remove_(self, integer_list, values_list):
        return [x for x in integer_list if x not in values_list]
```

Remove duplicate words

Ваша задача — удалить из строки все повторяющиеся слова, оставив только отдельные (первые) слова.

Пример:

Ввод:

'alpha beta beta gamma gamma gamma delta alpha beta beta gamma gamma gamma delta'

Вывод:

'alpha beta gamma delta'

```python
def remove_duplicate_words(s):
    words = s.split()
    result = []
    seen = set()

    for word in words:
        if word not in seen:
            result.append(word)
            seen.add(word)

    return ' '.join(result)
```
