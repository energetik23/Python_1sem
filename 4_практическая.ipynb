1. Переписать задания с разбиением на функции задания 2 и 4 из Практической 2.

```python
#задание 1
def calculate(operation: str, value_1: int, value_2: int):
    operations = {
        '+': lambda a, b: a + b,
        '-': lambda a, b: a - b,
        '*': lambda a, b: a * b,
    }
    if operation == '/':
        return value_1 / value_2 if value_2 != 0 else "Деление на ноль!"
    return operations.get(operation, lambda a, b: "Неизвестная операция")(value_1, value_2)


def get_quarter(month: int) -> int:
    return (month - 1) // 3 + 1


def get_season(month: int) -> str:
    seasons = {
        (12, 1, 2): "Зима",
        (3, 4, 5): "Весна",
        (6, 7, 8): "Лето",
        (9, 10, 11): "Осень",
    }
    for months, season in seasons.items():
        if month in months:
            return season
    return "Некорректный месяц"


def month_info(month: int) -> str:
    if not 1 <= month <= 12:
        return "Некорректный номер месяца"
    return f"Месяц {month}: {get_quarter(month)} квартал, {get_season(month)}"


print(calculate('+', 4, 7))   # 11
print(month_info(7))          # Месяц 7: 3 квартал, Лето
```

2. Написать функцию для подсчета определенных символов в предложении, записанных в словарь, вывести количество каждой.

```python
#задание 2
def count_chars(sentence: str) -> dict:
    result = {}
    for ch in sentence:
        result[ch] = result.get(ch, 0) + 1
    return result


text = "hello world"
print(count_chars(text))
# {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}
```

3. В строке содержатся только символы от a до z. Написать функцию вывода количества символов из диапазона от m до z.

```python
#задание 3
def count_m_to_z(text: str) -> int:
    return sum(1 for ch in text if 'm' <= ch <= 'z')


print(count_m_to_z("abcdefghijklmnopqrstuvwxyz"))  # 14 (m..z)
```
