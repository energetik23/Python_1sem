**Refactored Greeting**

В следующем коде можно было бы применить немного объектно-ориентированного подхода. Хотя это простой метод и он отлично работает сам по себе, в более крупной системе лучше организовать методы в классы/объекты. (Или, по крайней мере, сделать что-то подобное, в зависимости от вашего языка программирования)

Измените рефакторинг следующего кода так, чтобы он относился к классу/объекту Person. У каждого экземпляра Person будет метод greet. Экземпляр Person должен создаваться с именем, чтобы его не нужно было передавать при каждом вызове метода greet.

Вот как будет использоваться окончательный рефакторинг кода:

joe = Person('Joe')

joe.greet('Kate') # should return 'Hello Kate, my name is Joe'

joe.name          # should == 'Joe'

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self, your_name):
        return "Hello %s, my name is %s" % (your_name, self.name)
```

**Finish Guess the Number Game**

Представьте, что вы создаете игру, в которой пользователь должен угадать правильное число. Но количество попыток ограничено.

Если пользователь попытается угадать число больше одного раза, функция должна выдать ошибку.
Если пользователь угадает число, функция должна вернуть true.
Если пользователь угадал неправильно, функция должна вернуть false и лишить его жизни.
Сможете ли вы пройти игру так, чтобы соблюсти все правила?

```python
class Guesser:
    def __init__(self, number, lives):
        self.number = number
        self.lives = lives

    def guess(self, n):
        if self.lives <= 0:
            raise Exception("No lives left")

        if n == self.number:
            return True

        self.lives -= 1
        return False
```
