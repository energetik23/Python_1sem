1. На входе строка, которую нужно найти в списке строк, указав индекс найденного элемента. Выдать соответствующее сообщение при отсутствии данной строки в списке.

```python
def find_string(strings, target):
    if target in strings:
        index = strings.index(target)
        print(f"Строка '{target}' найдена по индексу {index}")
    else:
        print(f"Строка '{target}' отсутствует в списке")

# Пример
lst = ["яблоко", "банан", "вишня", "груша"]
find_string(lst, "вишня")
find_string(lst, "арбуз")
```

2. Удалить дубликаты элементов из списка.

```python
def remove_duplicates(lst):
    return list(dict.fromkeys(lst))

# Пример
nums = [1, 2, 2, 3, 1, 4, 3, 5]
print(remove_duplicates(nums))  # [1, 2, 3, 4, 5]
```

3. Найти количество повторений элементов в списке (использовать словарь).

```python
def count_occurrences(lst):
    counts = {}
    for item in lst:
        counts[item] = counts.get(item, 0) + 1
    return counts

# Пример
nums = [1, 2, 2, 3, 1, 4, 3, 3]
result = count_occurrences(nums)
for key, value in result.items():
    print(f"Элемент {key} встречается {value} раз(а)")
```

4. Найти и удалить из списка минимальное целое число, при этом не меняя порядок элементов.

```python
def remove_min(lst):
    if not lst:
        return lst
    minimum = min(lst)
    lst.remove(minimum)  # remove удаляет только первое вхождение
    return lst

# Пример
nums = [5, 3, 8, 1, 9, 1, 4]
print(remove_min(nums))  # [5, 3, 8, 9, 1, 4]
```
