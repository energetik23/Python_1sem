1. Рассчитать средний возраст n человек, n вводится с клавиатуры

```python
n = int(input("Введите количество человек: "))

s = 0
for i in range(n):
    age = int(input("Введите возраст: "))
    s += age

average = s / n

print("Средний возраст:", average)
```

2. На входе строка цветового кода (например "a0351f"), определить цвет ближе к красному, синему или зеленому. Вывести что цвет серый, если все значения равны, но цвет не черный и не белый. Вывести "черный" или "белый" если код 000000 или ffffff соответственно

```python
color = input("Введите цветовой код: ").lower()

r = int(color[0:2], 16)
g = int(color[2:4], 16)
b = int(color[4:6], 16)

if color == "000000":
    print("черный")
elif color == "ffffff":
    print("белый")
elif r == g == b:
    print("серый")
elif r >= g and r >= b:
    print("красный")
elif g >= r and g >= b:
    print("зеленый")
else:
    print("синий")
```
