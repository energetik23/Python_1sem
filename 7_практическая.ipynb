1. Создать базовый класс Shape с абстрактным методом area и выводом наименования фигуры через  __str__. Наследуемые классы Circle и Rectangle с методами вычисления площади, вывода параметров необходимых для вычисления площади и сравнения площади с другой фигурой (больше, меньше или равно) через __eq__.

```python

from abc import ABC, abstractmethod
import math


# Базовый абстракт класс
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    def __str__(self):
        return self.__class__.__name__


# Класс круга
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __str__(self):
        return f"Круг (радиус = {self.radius})"

    def __eq__(self, other):
        return self.area() == other.area()

    def __lt__(self, other):
        return self.area() < other.area()

    def __gt__(self, other):
        return self.area() > other.area()


# Класс прямоугольника
class Rectangle(Shape):

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def __str__(self):
        return f"Прямоугольник (ширина = {self.width}, высота = {self.height})"

    def __eq__(self, other):
        return self.area() == other.area()

    def __lt__(self, other):
        return self.area() < other.area()

    def __gt__(self, other):
        return self.area() > other.area()


# Создание фигур
circle = Circle(5)
rectangle = Rectangle(10, 8)

# Вывод информации о фигурах
print(circle)
print("Площадь круга:", round(circle.area(), 2))

print(rectangle)
print("Площадь прямоугольника:", rectangle.area())

# Сравнение площадей
print("\nСравнение фигур:")

if circle > rectangle:
    print("Площадь круга больше площади прямоугольника")
elif circle < rectangle:
    print("Площадь круга меньше площади прямоугольника")
else:
    print("Площади фигур равны")
```
