**Текст, выделенный полужирным шрифтом**
Часы показывают h часов, m минут и s секунд после полуночи.

Ваша задача — написать функцию, которая возвращает время, прошедшее с полуночи, в миллисекундах.

Пример:

h = 0

m = 1

s = 1

result = 61000

Ограничения на входные данные:

0 <= h <= 23

0 <= m <= 59

0 <= s <= 59

```python
def past(h, m, s):
    return h * 3600000 + m * 60000 + s * 1000
```

**Grasshopper - Array Mean**

Найти среднее значение

Найдите среднее значение (среднюю арифметическую) списка чисел в массиве.

Информация

Чтобы найти среднее арифметическое (среднее значение) набора чисел, сложите все числа и разделите на количество значений в списке.

Пример списка 1, 3, 5, 7

1. Сложите все числа

1+3+5+7 = 16

2. Разделите на количество значений в списке. В этом примере в списке четыре числа.

16/4 = 4

3. Среднее значение (или медиана) этого списка равно 4

```python
def find_average(nums):
    return sum(nums) / len(nums)
```

Задача

Для заданного целого числа определите, является ли оно квадратным числом:

 В математике квадратное число или полный квадрат — это целое число, являющееся квадратом целого числа; другими словами, это произведение некоторого целого числа на само себя.

В тестах всегда используется какое-либо целое число, так что не беспокойтесь об этом в языках с динамической типизацией.

Примеры

-1  =>  false

 0  =>  true

 3  =>  false

 4  =>  true

25  =>  true

26  =>  false

```python
def is_square(n):
    if n < 0:
        return False
    return int(n ** 0.5) ** 2 == n
```
