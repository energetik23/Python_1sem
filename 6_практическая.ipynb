1. Создать класс студента. Атрибуты: name, group, grades, tel_number. Методы: add_grade(grade), get_average(), info(). Использовать методы для добавления оценки студенту, получению среднего балла и всей информации соответствующими методами. Описать на основе созданного класса трех студентов. Номер телефона должен быть приватным атрибутом класса.

```python
class Student:
    def __init__(self, name: str, group: str, tel_number: str, grades=None):
        self.name = name
        self.group = group
        self.__tel_number = tel_number
        self.grades = grades if grades is not None else []

    def add_grade(self, grade):
        """Добавить оценку студенту."""
        if not isinstance(grade, (int, float)) or isinstance(grade, bool):
            raise TypeError("Оценка должна быть числом")
        if not (1 <= grade <= 5):
            raise ValueError("Оценка должна быть в диапазоне 1-5")
        self.grades.append(grade)

    def get_average(self):
        """Вернуть средний балл (0, если оценок нет)."""
        if not self.grades:
            return 0
        return round(sum(self.grades) / len(self.grades), 2)

    def info(self):
        """Вернуть строку со всей информацией о студенте."""
        return (f"Студент: {self.name} | Группа: {self.group} | "
                f"Оценки: {self.grades} | Средний балл: {self.get_average()} | "
                f"Телефон: {self.__tel_number}")


# Создаём 3 студентов
s1 = Student("Иванов Иван", "ИС-21", "+7-900-111-22-33", [4, 5])
s2 = Student("Петрова Анна", "ИС-21", "+7-900-444-55-66")
s3 = Student("Сидоров Пётр", "ПМ-22", "+7-900-777-88-99", [3, 4, 5])

# Добавляем оценки через метод
s1.add_grade(5)
s2.add_grade(4)
s2.add_grade(3)
s3.add_grade(4)

# Выводим инфу и ср балл
for s in (s1, s2, s3):
    print(s.info())
    print("Средний балл:", s.get_average())
    print("-" * 60)
```
