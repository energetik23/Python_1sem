**Classy Extensions**

Классные расширения

Цель этого задания — отработать базовые концепции ООП, связанные с наследованием и переопределением методов.

Вам будет предоставлен класс Animal, поэтому вам нужно будет отредактировать только класс Cat.

Задание
Ваша задача — дополнить класс Cat, который расширяет Animal и заменяет метод speak на тот, который возвращает имя кошки + «мяу». Например, 'Mr Whiskers meows.'

Атрибут name доступен в классе с self.name.

```python
from preloaded import Animal

class Cat(Animal):
    def speak(self):
        return self.name + " meows."
```

**Color Ghost**

Цветной призрак

Создайте класс Ghost

Объекты Ghost создаются без каких-либо аргументов.

При создании объектам Ghost присваивается случайный цветовой атрибут: "белый", "желтый", "фиолетовый" или "красный"

ghost = Ghost()

ghost.color  #=> "white" or "yellow" or "purple" or "red"

```python
import random

class Ghost:
    def __init__(self):
        self.color = random.choice(["white", "yellow", "purple", "red"])
```
