1. Перевести строкое значение в числовое и обратно, отобразить тип данных (ввод с клавиатуры)

```python
# Задание 1. Перевести строковое значение в числовое и обратно, отобразить тип данных (ввод с клавиатуры)

# Ввод строки с клавиатуры
user_input = input("Введите число: ")

# Перевод с проверкой
try:
    number_value = int(user_input)
except ValueError:
    print("Ошибка: введите число!")
    exit()

# Ошибки нет — выводим результат
print(f"Числовое значение: {number_value}, тип: {type(number_value)}")

# Перевод числа обратно в строку
string_value = str(number_value)
print(f"Строковое значение: '{string_value}', тип: {type(string_value)}")
```

2. Вернуть отрицательное число, введенное с клавиатуры. Если число уже отрицательное - вернуть исходное.

```python
# Задание 2. Вернуть отрицательное число, введенное с клавиатуры. Если число уже отрицательное - вернуть исходное.

# Ввод числа с проверкой
try:
    num = float(input("Введите число: "))
except ValueError:
    print("Ошибка: введите число!")
    exit()

if num > 0:
    result = -num
else:
    result = num

print(f"Результат: {result}")
```

3. На входе три переменные: cap - вместимость автобуса для пасссажиров, on - сколько пассажиров уже находится в автобусе, wait - число ожидающих посадку пассажиров. Необходимо определить, хватит ли места в автобусе для всех пассажиров, вернув число оставшихся ожидать пассажиров. Например: cap = 10, on = 5, wait = 5 --> 0 оставшихся пассажиров. cap = 100, on = 60, wait = 50 --> 10 оставшихся пассажиров.

```python
# Задание 3. На входе три переменные: cap - вместимость автобуса для пассажиров,on - сколько пассажиров уже находится в автобусе, wait - число ожидающих
# посадку пассажиров. Необходимо определить, хватит ли места в автобусе для всех пассажиров, вернув число оставшихся ожидать пассажиров.
# Например: cap = 10, on = 5, wait = 5 --> 0 оставшихся пассажиров. cap = 100, on = 60, wait = 50 --> 10 оставшихся пассажиров.

# Ввод данных с проверкой
try:
    cap = int(input("Введите вместимость автобуса (cap): "))
    on = int(input("Введите кол-во пассажиров в автобусе (on): "))
    wait = int(input("Введите кол-во ожидающих пассажиров (wait): "))
except ValueError:
    print("Ошибка: введите число!")
    exit()

# Считаем свободные места
free_seats = cap - on

if free_seats >= wait:
    remaining = 0
else:
    remaining = wait - free_seats

print(f"Осталось ожидать пассажиров: {remaining}")
```

5. Вернуть произведение двух чисел, введенных с клавиатуры

```python
# Задание 5. Вернуть произведение двух чисел, введенных с клавиатуры

# Ввод 1 числа с проверкой
try:
    num1 = float(input("Введите первое число: "))
except ValueError:
    print("Ошибка введите число")
    exit()

# Ввод 2 числа с проверкой
try:
    num2 = float(input("Введите второе число: "))
except ValueError:
    print("Ошибка введите число")
    exit()

# Результат
product = num1 * num2
print(f"Произведение чисел: {product}")
```
